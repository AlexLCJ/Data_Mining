2.考拉茨猜想，对每一个正整数，如果它是奇数，则对它乘3再加1，如果它是偶数，
则对它除以2，如此循环，最后都能得到1。编写一个程序，输入一个正整数，
验证考拉茨猜想并打印考拉茨序列。例如，输入6，从6开始的考拉茨序列是6，3，
10，5，16，8，4，2，1 （20分）

```python
# 代码
number = int(input("输入一个整数"))
def sequence(n):
    sequence = [n]
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        sequence.append(n)
    return sequence
print(sequence(number))
```

3. 求10的阶乘，要求利用递归函数（20分）

```python
# 代码
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
print(factorial(10))
```

|4. 已知若干学生的成绩按行存储在文件a.txt(按要求自己建立)中，每一行表示一个学生的基本信息和成绩信息，其中成绩均为百分制，以英文逗号为分隔符。请计
算每个学生的综合成绩，并按综合成绩降序排列，写入到文件b.txt中。
综合成绩计算公式：平时成绩*0.2+实验成绩*0.2+期末考试成绩*0.6。
文件格式为：
   a.txt：学号,姓名,平时成绩,实验成绩,期末考试成绩
   b.txt：学号,姓名,平时成绩,实验成绩,期末考试成绩,综合成绩

```python
# 代码
with open('a.txt', 'r', encoding='utf-8') as file:
    lines = file.readlines()
# print(lines)
students = []
for line in lines:
      data = line.strip().split(',')
      students.append({
          '学号': data[0],
          '姓名': data[1],
          '平时成绩': float(data[2]),
          '实验成绩': float(data[3]),
          '期末考试成绩': float(data[4])
      })
      students.sort(key=lambda x: x['平时成绩'] * 0.2 +
                                  x['实验成绩'] * 0.2 + x['期末考试成绩'] * 0.6, 
                    reverse=True)
      with open('b.txt', 'w', encoding='utf-8') as output_file:
          for student in students:
              output_file.write(f"{student['学号']},{student['姓名']},"
                                f"{student['平时成绩']},{student['实验成绩']},"
                                f"{student['期末考试成绩']},"
                                f"{student['平时成绩'] * 0.2 + student['实验成绩'] * 0.2 + student['期末考试成绩'] * 0.6}\n")
```
