```python
#1（20）.函数shift的功能是将列表lst的元素循环左移二位。
#   例如：输入[2,3,4,5,6,7] ，输出[3,4,5,6,7,2]，纠正程序中的错误
def shift(lst):
    temp=lst[0]               
    for i in range(len(lst)):  
        lst[i-1]==lst[i]       
    lst[len(lst)-1]=temp   
lst = [2,3,4,5,6,7]
print(shift(lst))
```

```python
# 纠正：
def shift(lst):
    n = len(lst)
    if n <= 2:
        return lst  
    temp = lst[:2]  
    lst = lst[2:] + temp 
    return lst
lst = [2, 3, 4, 5, 6, 7]
print(shift(lst))
```

```python
# 2（20）函数reverse的功能是将列表a的元素反转存放。
# 例如： 输入：[1,12,3,45] ， 输出：[45,3,12,1]
def reverse(a):
    t=len(a)
    i=0            
    while i < t:
        j = t-i-1    
        if i<j:
            a[i],a[j] = a[j],a[i]  
        i=i+1
listA = [12,3,56,78,100,23]
reverse(listA)
print(listA)
```

```python
# 3（20）找出1:100的能被4和9整除的数，并找出与7互素的数
def find_numbers():
    numbers = []
    for num in range(1, 101):
        if num % 4 == 0 and num % 9 == 0:
            numbers.append(num)
    for num in numbers:
        if num % 7 != 0:
            print(num)
find_numbers()
```

```python
def f(i):
    """计算函数f(i)的值"""
    """
    计算从 1 到 i 的每个整数 j 与 j+1 的比值之和
    """
    result = 0.0
    for j in range(1, i + 1):
        result += j / (j + 1)
    return result
 
def main():
    try:
        i = int(input("请输入一个正整数 i: "))
        if i <= 0:
            raise ValueError("i 必须是一个正整数")
        result = f(i)
        print(f"f({i}) = {result}")
    except ValueError as e:
        print(f"输入错误: {e}")
 
if __name__ == "__main__":
    main()
```
