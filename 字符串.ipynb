### 1.统计字符串中每个字符连续出现的次数
编写一个Python程序，从键盘接收一个字符串，统计其中每个字符连续出现的次数，并将其重构为一种新的格式，格式为“字符+出现次数”。如果字符连续出现一次，则只保留字符本身。\
例如，如果输入字符串为 "aaabbcaaa", 则输出结果为 "a3b2ca3"。

```python
input_str = input("请输入一个字符串：")

result = ""
count = 1

for i in range(1, len(input_str)):
    # 如果当前字符与前一个字符相同，计数器加1
    if input_str[i] == input_str[i - 1]:
        count += 1
    else:
        # 否则，将前一个字符和计数添加到结果字符串中
        result += input_str[i - 1] + (str(count) if count > 1 else "")
        count = 1  # 重置计数器

# 添加最后一个字符及其计数
result += input_str[-1] + (str(count) if count > 1 else "")

print("压缩后的字符串：", result)
```

### 2.判断字符串中的括号是否有效配对

编写一个Python程序，判断输入的字符串中各种类型的括号 (), {}, [] 是否有效配对。括号的配对需要满足以下条件：

每个左括号必须有一个相同类型的右括号与之对应。
括号之间的嵌套结构必须正确。

例如，字符串 "(a + b) * [c - {d / e}]" 是有效的，而 "(a + b] * {c - d}" 是无效的

```python
def is_valid_brackets(s):
    stack = []  # 栈思想
    bracket_pairs = {')': '(', '}': '{', ']': '['}  # 配对字典

    for char in s:
        if char in bracket_pairs.values():
            # 如果是左括号，压入栈中
            stack.append(char)
        elif char in bracket_pairs:
            # 如果是右括号，检查栈顶是否匹配
            if not stack or stack[-1] != bracket_pairs[char]:
                return False
            stack.pop()  # 匹配则弹出栈顶元素

    # 检查栈是否为空，若为空则表示完全匹配
    return len(stack) == 0


input_str = input("请输入一个包含括号的字符串：")
is_valid = is_valid_brackets(input_str)
print("括号是否有效配对：", is_valid)
```

### 3.最长公共前缀
from:leetcode第14题


编写一个函数来查找字符串数组中的最长公共前缀。

如果不存在公共前缀，返回空字符串 ""。

示例：\
输入：strs = ["flower","flow","flight"]\
输出："fl"

```python
# 输入列表形式的字符串，例如 ["flower", "flow", "flight"]
List = eval(input("请输入字符串列表，例如 [\"flower\", \"flow\", \"flight\"]: "))

class Solution:
    def longestCommonPrefix(self, strs: list[str]) -> str:
        if not strs:
            return ""
        
        # 找到字典序最小和最大的字符串
        str0 = min(strs)
        str1 = max(strs)
        
        # 比较两个字符串，找到最长公共前缀
        for i in range(len(str0)):
            if str0[i] != str1[i]:
                return str0[:i]
        
        return str0

# 创建 Solution 类的实例并调用方法
solution = Solution()
result = solution.longestCommonPrefix(List)
print("最长公共前缀是:", result)
```
