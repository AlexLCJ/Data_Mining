```python
fname = input("请输入要写入的文件：")
fo = open(fname, "w+", encoding="utf-8")  # 使用 utf-8 编码以正确处理中文字符
ls = ["床前明月光，\n", "疑是地上霜。\n",
      "举头望明月，\n", "低头思故乡。\n"]
fo.writelines(ls)  # 正确使用 writelines() 方法写入行列表
 
# 将文件指针移动到文件开头以便读取
fo.seek(0)
 
# 读取并打印文件内容
for line in fo:
    print(line, end='')  # 使用 end='' 防止打印额外的空行
 
fo.close()
```
